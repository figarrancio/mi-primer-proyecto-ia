```python
from transformers import pipeline

# Descargamos un modelo pequeño de clasificación de sentimientos en español
analizador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis"
)

# Probemos con dos frases
frases = [
    "El servicio se restableció con éxito y la red está estable.",
    "Hubo una degradación crítica en el enlace principal."
]

resultados = analizador(frases)

for frase, res in zip(frases, resultados):
    print(f"Frase: {frase}")
    print(f"Diagnóstico IA: {res['label']} (Confianza: {res['score']:.2%})\n")
```
